```python
# create function replace null data

def replaceNullData(rows):
    for col in rows:
        if rows[col].dtype == 'object' :
            rows[col].fillna(rows[col].mode()[0],inplace=True)
        else:
            rows[col].fillna(rows[col].mean(),inplace=True)
            
    return rows
```

```python
# function for Outliers
def outliners(rows):

    # Process each column
    for col in rows.columns:

        # Process only numeric columns
        if pd.api.types.is_numeric_dtype(rows[col]):

            Q1 = rows[col].quantile(0.25)
            Q3 = rows[col].quantile(0.75)

            IQR = Q3 - Q1

            lower_bound = Q1 - (1.5 * IQR)
            upper_bound = Q3 + (1.5 * IQR)

            mean_value = rows[col].mean()

            # Count outliers
            outlier_count = (
                (rows[col] < lower_bound) |
                (rows[col] > upper_bound)
            ).sum()

            # Replace outliers with mean
            rows.loc[rows[col] < lower_bound, col] = mean_value
            rows.loc[rows[col] > upper_bound, col] = mean_value

            print(f"{col}: {outlier_count} outliers replaced")

    return rows
```

```python
def decile_analysis(rows):

    # Step 1: Aggregate data per user
    user_data = rows.groupby("MSISDN/Number").agg({
        "Dur. (ms)": "sum",
        "Total DL (Bytes)": "sum",
        "Total UL (Bytes)": "sum"
    }).reset_index()

    # Step 2: Calculate total data (DL + UL)
    user_data["Total Data (DL + UL)"] = (
        user_data["Total DL (Bytes)"] +
        user_data["Total UL (Bytes)"]
    )

    # Step 3: Create 10 decile classes based on total duration
    user_data["Decile"] = pd.qcut(
        user_data["Dur. (ms)"],
        10,
        labels=False,
        duplicates="drop"
    )

    # Step 4: Convert 0-9 into Decile 1-10
    user_data["Decile"] = user_data["Decile"] + 1

    # Step 5: Select top 5 deciles
    top_5_deciles = user_data[
        user_data["Decile"] >= 6
    ]

    # Step 6: Calculate total data for each decile
    result = (
        top_5_deciles
        .groupby("Decile")
        .agg({
            "Dur. (ms)": "sum",
            "Total Data (DL + UL)": "sum"
        })
        .reset_index()
    )

    return user_data, result
```

```python
def strongest_correlations(rows):

    columns = [
        "Social Media",
        "Google",
        "Email",
        "YouTube",
        "Netflix",
        "Gaming",
        "Other"
    ]

    corr = rows[columns].corr()

    mask = np.triu(
        np.ones(corr.shape),
        k=1
    ).astype(bool)

    result = corr.where(mask).stack().reset_index()

    result.columns = [
        "Application 1",
        "Application 2",
        "Correlation"
    ]

    result["Absolute Correlation"] = result["Correlation"].abs()

    return result.sort_values(
        "Absolute Correlation",
        ascending=False
    ).reset_index(drop=True)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

def perform_pca(rows):

    # Application variables
    application_columns = [
        "Social Media",
        "Google",
        "Email",
        "YouTube",
        "Netflix",
        "Gaming",
        "Other"
    ]

    # Select data
    X = rows[application_columns].copy()

    # Remove missing values
    X = X.dropna()

    # Standardize data
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Apply PCA
    pca = PCA()
    principal_components = pca.fit_transform(X_scaled)

    # Explained variance
    pca_result = pd.DataFrame({
        "Principal Component": [
            f"PC{i+1}" for i in range(len(pca.explained_variance_ratio_))
        ],
        "Explained Variance": pca.explained_variance_ratio_,
        "Cumulative Variance": pca.explained_variance_ratio_.cumsum()
    })

    return pca, principal_components, pca_result
```

```python
def top_10_users_per_application(data):

    applications = {
        "Social Media": (
            "Social Media DL (Bytes)",
            "Social Media UL (Bytes)"
        ),
        "Google": (
            "Google DL (Bytes)",
            "Google UL (Bytes)"
        ),
        "Email": (
            "Email DL (Bytes)",
            "Email UL (Bytes)"
        ),
        "YouTube": (
            "Youtube DL (Bytes)",
            "Youtube UL (Bytes)"
        ),
        "Netflix": (
            "Netflix DL (Bytes)",
            "Netflix UL (Bytes)"
        ),
        "Gaming": (
            "Gaming DL (Bytes)",
            "Gaming UL (Bytes)"
        ),
        "Other": (
            "Other DL (Bytes)",
            "Other UL (Bytes)"
        )
    }

    # Aggregate application traffic per customer
    user_application = data.groupby(
        "MSISDN/Number"
    )[
        [col for app in applications.values() for col in app]
    ].sum()

    # Store top 10 results
    top_users = {}

    for app, (dl_col, ul_col) in applications.items():

        # Total DL + UL for application
        user_application[app] = (
            user_application[dl_col] +
            user_application[ul_col]
        )

        # Top 10 users
        top_users[app] = (
            user_application[app]
            .nlargest(10)
            .reset_index()
        )

        top_users[app].columns = [
            "MSISDN/Number",
            "Total Traffic (Bytes)"
        ]

    return user_application, top_users
```

```python
def combined_top_users(top_users):

    result = []

    for app, df in top_users.items():

        temp = df.copy()
        temp["Application"] = app

        result.append(temp)

    return pd.concat(
        result,
        ignore_index=True
    )[[
        "Application",
        "MSISDN/Number",
        "Total Traffic (Bytes)"
    ]]
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

def find_optimal_k(customer_data, max_k=10):

    engagement_cols = [
        "Number of Sessions",
        "Total Session Duration (ms)",
        "Total Traffic (Bytes)"
    ]

    # Select engagement metrics
    X = customer_data[engagement_cols].dropna()

    # Normalize the data
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Calculate WCSS for different k values
    wcss = []

    for k in range(1, max_k + 1):

        kmeans = KMeans(
            n_clusters=k,
            random_state=42,
            n_init=10
        )

        kmeans.fit(X_scaled)

        wcss.append(kmeans.inertia_)

    # Plot Elbow curve
    plt.figure(figsize=(8, 5))

    plt.plot(
        range(1, max_k + 1),
        wcss,
        marker="o"
    )

    plt.title("Elbow Method for Optimal K")
    plt.xlabel("Number of Clusters (K)")
    plt.ylabel("WCSS / Inertia")
    plt.xticks(range(1, max_k + 1))

    plt.tight_layout()
    plt.show()

    return wcss
```

```python
wcss = find_optimal_k(customer_data)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
def engagement_clustering(customer_data, k=3):

    engagement_cols = [
        "Number of Sessions",
        "Total Session Duration (ms)",
        "Total Traffic (Bytes)"
    ]

    X = customer_data[engagement_cols].dropna()

    # Normalize
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # K-Means
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    clusters = kmeans.fit_predict(X_scaled)

    # Add cluster to the corresponding customers
    customer_data = customer_data.loc[X.index].copy()
    customer_data["Engagement Cluster"] = clusters

    return customer_data, kmeans, scaler
```

```python
def label_engagement_clusters(customer_data):

    metrics = [
        "Number of Sessions",
        "Total Session Duration (ms)",
        "Total Traffic (Bytes)"
    ]

    summary = customer_data.groupby(
        "Engagement Cluster"
    )[metrics].mean()

    # Average rank across the three metrics
    summary["Rank"] = summary.rank().mean(axis=1)

    sorted_clusters = summary["Rank"].sort_values().index

    labels = {}

    if len(sorted_clusters) == 3:
        labels = {
            sorted_clusters[0]: "Low Engagement",
            sorted_clusters[1]: "Medium Engagement",
            sorted_clusters[2]: "High Engagement"
        }

    customer_data["Engagement Level"] = (
        customer_data["Engagement Cluster"].map(labels)
    )

    return customer_data, summary
```

```python
## Task 3. 1 - Aggregate, per customer, the following information (treat missing & outliers by replacing with the mean or the mode of the corresponding variable):
## ● Average TCP retransmission
## ● Average RTT
## ● Handset type
## ● Average throughput
```

```python
def task_3_1(rows):

    # -------------------------------------------------
    # 1. Replace missing values
    # -------------------------------------------------

    numeric_cols = [
        "TCP DL Retrans. Vol (Bytes)",
        "TCP UL Retrans. Vol (Bytes)",
        "Avg RTT DL (ms)",
        "Avg RTT UL (ms)",
        "Avg Bearer TP DL (kbps)",
        "Avg Bearer TP UL (kbps)"
    ]

    for col in numeric_cols:
        rows[col] = rows[col].fillna(rows[col].mean())

    # Handset Type - missing values with mode
    mode_handset = rows["Handset Type"].mode()[0]

    rows["Handset Type"] = rows["Handset Type"].fillna(mode_handset)

    # -------------------------------------------------
    # 2. Treat outliers using mean
    # -------------------------------------------------

    for col in numeric_cols:

        Q1 = rows[col].quantile(0.25)
        Q3 = rows[col].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        mean_value = rows[col].mean()

        rows.loc[
            (rows[col] < lower_bound) |
            (rows[col] > upper_bound),
            col
        ] = mean_value

    # -------------------------------------------------
    # 3. Create average TCP retransmission
    # -------------------------------------------------

    rows["Average TCP Retransmission"] = (
        rows["TCP DL Retrans. Vol (Bytes)"] +
        rows["TCP UL Retrans. Vol (Bytes)"]
    ) / 2

    # -------------------------------------------------
    # 4. Create average RTT
    # -------------------------------------------------

    rows["Average RTT"] = (
        rows["Avg RTT DL (ms)"] +
        rows["Avg RTT UL (ms)"]
    ) / 2

    # -------------------------------------------------
    # 5. Create average throughput
    # -------------------------------------------------

    rows["Average Throughput"] = (
        rows["Avg Bearer TP DL (kbps)"] +
        rows["Avg Bearer TP UL (kbps)"]
    ) / 2

    # -------------------------------------------------
    # 6. Aggregate per customer
    # -------------------------------------------------

    customer_data = rows.groupby(
        "MSISDN/Number"
    ).agg({
        "Average TCP Retransmission": "mean",
        "Average RTT": "mean",
        "Handset Type": lambda x: x.mode()[0],
        "Average Throughput": "mean"
    }).reset_index()

    return customer_data
```

```python
def label_experience_clusters(experience_data):

    metrics = [
        "Average TCP Retransmission",
        "Average RTT",
        "Average Throughput"
    ]

    summary = experience_data.groupby(
        "Experience Cluster"
    )[metrics].mean()

    # Convert metrics into comparable ranks
    # Lower TCP = better
    # Lower RTT = better
    # Higher Throughput = better

    tcp_rank = summary[
        "Average TCP Retransmission"
    ].rank(ascending=True)

    rtt_rank = summary[
        "Average RTT"
    ].rank(ascending=True)

    throughput_rank = summary[
        "Average Throughput"
    ].rank(ascending=False)

    summary["Experience Score"] = (
        tcp_rank +
        rtt_rank +
        throughput_rank
    )

    # Lower score = better experience
    sorted_clusters = summary[
        "Experience Score"
    ].sort_values().index

    labels = {
        sorted_clusters[0]: "Good Experience",
        sorted_clusters[1]: "Average Experience",
        sorted_clusters[2]: "Poor Experience"
    }

    experience_data["Experience Level"] = (
        experience_data["Experience Cluster"]
        .map(labels)
    )

    return experience_data, summary
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from scipy.spatial.distance import cdist

def calculate_engagement_experience_scores(
    customer_data,
    experience_data,
    engagement_kmeans,
    engagement_scaler,
    experience_kmeans,
    experience_scaler
):

    # =========================================================
    # 1. ENGAGEMENT SCORE
    # =========================================================

    engagement_cols = [
        "Number of Sessions",
        "Total Session Duration (ms)",
        "Total Traffic (Bytes)"
    ]

    # Standardized engagement data
    engagement_X = customer_data[engagement_cols].copy()

    engagement_scaled = engagement_scaler.transform(
        engagement_X
    )

    # Get average values of each engagement cluster
    engagement_cluster_means = customer_data.groupby(
        "Engagement Cluster"
    )[engagement_cols].mean()

    # Find least engaged cluster
    # Rank each metric: lower = less engagement
    engagement_rank = (
        engagement_cluster_means.rank(ascending=True)
        .mean(axis=1)
    )

    least_engaged_cluster = engagement_rank.idxmin()

    print("Least Engaged Cluster:",
          least_engaged_cluster)

    # Get centroid from K-Means
    least_engaged_centroid = engagement_kmeans.cluster_centers_[
        list(engagement_kmeans.labels_).index(
            engagement_kmeans.labels_[
                list(customer_data["Engagement Cluster"]).index(
                    least_engaged_cluster
                )
            ]
        )
    ]

    # Calculate Euclidean distance
    engagement_scores = cdist(
        engagement_scaled,
        least_engaged_centroid.reshape(1, -1),
        metric="euclidean"
    ).flatten()

    customer_data = customer_data.copy()

    customer_data["Engagement Score"] = engagement_scores


    # =========================================================
    # 2. EXPERIENCE SCORE
    # =========================================================

    experience_cols = [
        "Average TCP Retransmission",
        "Average RTT",
        "Average Throughput"
    ]

    experience_X = experience_data[experience_cols].copy()

    experience_scaled = experience_scaler.transform(
        experience_X
    )

    # Calculate average metrics for each cluster
    experience_cluster_means = experience_data.groupby(
        "Experience Cluster"
    )[experience_cols].mean()

    # For experience:
    # Lower TCP = better
    # Lower RTT = better
    # Higher throughput = better

    tcp_rank = experience_cluster_means[
        "Average TCP Retransmission"
    ].rank(ascending=True)

    rtt_rank = experience_cluster_means[
        "Average RTT"
    ].rank(ascending=True)

    throughput_rank = experience_cluster_means[
        "Average Throughput"
    ].rank(ascending=False)

    experience_score_rank = (
        tcp_rank +
        rtt_rank +
        throughput_rank
    )

    # Worst experience cluster
    worst_experience_cluster = (
        experience_score_rank.idxmax()
    )

    print("Worst Experience Cluster:",
          worst_experience_cluster)

    # Get worst cluster centroid
    worst_experience_centroid = (
        experience_kmeans.cluster_centers_[
            worst_experience_cluster
        ]
    )

    # Euclidean distance
    experience_scores = cdist(
        experience_scaled,
        worst_experience_centroid.reshape(1, -1),
        metric="euclidean"
    ).flatten()

    experience_data = experience_data.copy()

    experience_data["Experience Score"] = experience_scores


    return (
        customer_data,
        experience_data,
        least_engaged_cluster,
        worst_experience_cluster
    )
```

```python
import mlflow
import mlflow.sklearn
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mlflow.set_experiment("Telecom_Satisfaction_Model")
```

```python
from sklearn.linear_model import LinearRegression

with mlflow.start_run(run_name="Satisfaction_Regression_Model"):

    # Start time is automatically recorded by MLflow

    # Parameters
    model_type = "Linear Regression"
    test_size = 0.20
    random_state = 42

    # Train model
    model = LinearRegression()

    model.fit(
        X_train,
        y_train
    )

    # Prediction
    y_pred = model.predict(X_test)

    # Metrics
    mae = mean_absolute_error(
        y_test,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            y_pred
        )
    )

    r2 = r2_score(
        y_test,
        y_pred
    )

    # Log parameters
    mlflow.log_param(
        "model_type",
        model_type
    )

    mlflow.log_param(
        "test_size",
        test_size
    )

    mlflow.log_param(
        "random_state",
        random_state
    )

    mlflow.log_param(
        "number_of_features",
        X_train.shape[1]
    )

    # Log metrics
    mlflow.log_metric(
        "MAE",
        mae
    )

    mlflow.log_metric(
        "RMSE",
        rmse
    )

    mlflow.log_metric(
        "R2",
        r2
    )

    # Log model
    mlflow.sklearn.log_model(
        model,
        "satisfaction_model"
    )

    print("Model tracked successfully!")

    print("MAE :", mae)
    print("RMSE:", rmse)
    print("R2  :", r2)
```
